```python
query1 = """
SELECT "Category", "Sales", "Profit"
FROM sales
WHERE "Category" = 'Technology'
LIMIT 10
"""
result1 = pd.read_sql_query(query1, conn)
print(result1)
```

```python
query2 = """
SELECT "Category", SUM("Sales") as Total_Sales, SUM("Profit") as Total_Profit
FROM sales
GROUP BY "Category"
ORDER BY Total_Sales DESC
"""
result2 = pd.read_sql_query(query2, conn)
print(result2)
```

```python
query3 = """
SELECT * 
FROM sales
WHERE "Category" = 'Furniture' 
  AND "Discount" > 0.3
"""
result3 = pd.read_sql_query(query3, conn)
print(result3)
print("\nعدد الصفوف:", len(result3))
```

```python
query4 = """
SELECT 
    COUNT(*) as Total_Orders,
    SUM(CASE WHEN "Profit" < 0 THEN 1 ELSE 0 END) as Losing_Orders,
    AVG("Profit") as Average_Profit
FROM sales
WHERE "Category" = 'Furniture' 
  AND "Discount" > 0.3
"""
result4 = pd.read_sql_query(query4, conn)
print(result4)
```

```python
query5 = """
SELECT "Order.ID", "Category", "Profit"
FROM sales
WHERE "Profit" > (SELECT AVG("Profit") FROM sales)
LIMIT 10
"""
result5 = pd.read_sql_query(query5, conn)
print(result5)
```

```python
import pandas as pd
import sqlite3

conn = sqlite3.connect('../data/superstore.db')
print("تم الاتصال بقاعدة البيانات بنجاح")
```

```python
query6 = """
WITH avg_profit AS (
    SELECT AVG("Profit") as avg_p
    FROM sales
)
SELECT "Order.ID", "Category", "Profit"
FROM sales, avg_profit
WHERE "Profit" > avg_p
LIMIT 10
"""
result6 = pd.read_sql_query(query6, conn)
print(result6)
```

```python
query7 = """
WITH category_margins AS (
    SELECT 
        "Category",
        SUM("Sales") as total_sales,
        SUM("Profit") as total_profit,
        (SUM("Profit") * 1.0 / SUM("Sales")) * 100 as profit_margin
    FROM sales
    GROUP BY "Category"
)
SELECT *
FROM category_margins
WHERE profit_margin < 10
"""
result7 = pd.read_sql_query(query7, conn)
print(result7)
```

```python
query_region = """
WITH region_sales AS (
    SELECT 
        "Region",
        AVG("Sales") as avg_sales
    FROM sales
    GROUP BY "Region"
)
SELECT *
FROM region_sales
WHERE avg_sales > 250
"""
result_region = pd.read_sql_query(query_region, conn)
print(result_region)
```

```python
query8 = """
SELECT 
    "Category",
    "Product.Name",
    "Profit",
    RANK() OVER (PARTITION BY "Category" ORDER BY "Profit" DESC) as profit_rank
FROM sales
LIMIT 20
"""
result8 = pd.read_sql_query(query8, conn)
print(result8)
```

```python
query9 = """
SELECT 
    "Order.Date",
    "Sales",
    SUM("Sales") OVER (ORDER BY "Order.Date") as running_total
FROM sales
ORDER BY "Order.Date"
LIMIT 15
"""
result9 = pd.read_sql_query(query9, conn)
print(result9)
```

```python
query9_fixed = """
SELECT 
    "Order.Date",
    "Order.ID",
    "Sales",
    SUM("Sales") OVER (ORDER BY "Order.Date", "Order.ID") as running_total
FROM sales
ORDER BY "Order.Date"
LIMIT 15
"""
result9_fixed = pd.read_sql_query(query9_fixed, conn)
print(result9_fixed)
```

```python
import pandas as pd
import sqlite3

conn = sqlite3.connect('../data/superstore.db')
```

```python
query10 = """
SELECT 
    "Customer.Name",
    SUM("Sales") AS Total_Sales,
    COUNT("Order.ID") AS Total_Orders
FROM 
    sales
GROUP BY 
    "Customer.Name"
ORDER BY 
    Total_Sales DESC
LIMIT 5
"""
result10 = pd.read_sql_query(query10, conn)
print(result10)
```

```python
query11 = """
SELECT 
    "Sub.Category",
    SUM("Sales") as Total_Sales,
    SUM("Quantity") as Total_Quantity
FROM sales
GROUP BY "Sub.Category"
ORDER BY Total_Sales DESC
LIMIT 5
"""
result11 = pd.read_sql_query(query11, conn)
print(result11)
```

```python
query11 = """
SELECT 
    "Ship.Mode",
    AVG("Shipping.Cost") as Avg_Shipping_Cost
FROM sales
GROUP BY "Ship.Mode"
ORDER BY Avg_Shipping_Cost DESC
"""
result11 = pd.read_sql_query(query11, conn)
print(result11)
```

```python
query12 = """
SELECT 
    strftime('%Y', "Order.Date") as Year,
    strftime('%m', "Order.Date") as Month,
    SUM("Sales") as Monthly_Sales
FROM sales
GROUP BY Year, Month
ORDER BY Monthly_Sales DESC
LIMIT 5
"""
result12 = pd.read_sql_query(query12, conn)
print(result12)
```

```python
query13 = """
SELECT "Market", COUNT(*) as Order_Count, SUM("Sales") as Total_Sales
FROM sales
GROUP BY "Market"
HAVING Order_Count > 5000
"""
result13 = pd.read_sql_query(query13, conn)
print(result13)
```

```python
query14 = """
SELECT 
    "Order.Priority",
    COUNT(*) as Order_Count,
    AVG("Profit") as Avg_Profit
FROM sales
GROUP BY "Order.Priority"
ORDER BY Avg_Profit DESC
"""
result14 = pd.read_sql_query(query14, conn)
print(result14)
```
